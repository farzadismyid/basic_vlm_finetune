```python
from src.models.qwen2vl_loader import load_qwen2vl
from src.models.inference import run_inference
from src.data.image_loader import load_image_from_url

model, processor = load_qwen2vl()
```

```python
image_url = "https://images.unsplash.com/photo-1529139574466-a303027c1d8b"

image = load_image_from_url(image_url)

image
```

```python
prompts = [
    "Describe this outfit professionally.",
    "Identify fashion style and aesthetic.",
    "Describe colors and clothing items.",
    "Would this outfit fit luxury fashion branding?",
]
```

```python
for prompt in prompts:

    result = run_inference(
        model=model,
        processor=processor,
        image=image,
        prompt=prompt,
        max_new_tokens=48,
    )

    print("=" * 60)
    print("PROMPT:", prompt)
    print("=" * 60)

    print(result["response"])

    print("\nTIME:", result["time_seconds"])
    print("\n")
```
